```python
import pandas as pd
df = pd.read_csv("data.csv")
df.head()
```

```python
df.isnull().sum()
```

```python
df.describe()
```

```python
df = df.drop('Unnamed: 32', axis = 1, errors = 'ignore')
df = df.drop('id', axis=1,  errors = 'ignore')
df
```

```python
df['diagnosis'] = df['diagnosis'].map({'M': 1, 'B': 0})
```

```python
df
```

```python
from sklearn.model_selection import train_test_split
x = df.drop('diagnosis', axis = 1)
y = df['diagnosis']
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = 0.2, random_state = 42)
x_train
```

```python
x_test
```

```python
y_train
```

```python
#Random Forest
from sklearn.ensemble import RandomForestClassifier
ran = RandomForestClassifier()

ran.fit(x_train,y_train)
ran.predict(x_test)
```

```python
score = ran.score(x_test,y_test)
score
```

```python
#Logistic Regressor
from sklearn.linear_model import LogisticRegression
log = LogisticRegression(max_iter=10000)

log.fit(x_train,y_train)
log.predict(x_test)
score2 = log.score(x_test,y_test)
score2
```

```python
#K-Nearest Neighbour
from sklearn.neighbors import KNeighborsClassifier
knn = KNeighborsClassifier() 
knn.fit(x_train,y_train)
knn.predict(x_test)
score3 = knn.score(x_test,y_test)
score3
```

```python
from sklearn.metrics import confusion_matrix

# Get the actual predicted answers from both modelss
log_predictions = log.predict(x_test)
knn_predictions = knn.predict(x_test)

print("--- LOGISTIC REGRESSION MISTAKES ---")
print(confusion_matrix(y_test, log_predictions))

print("\n--- KNN MISTAKES ---")
print(confusion_matrix(y_test, knn_predictions))
```
